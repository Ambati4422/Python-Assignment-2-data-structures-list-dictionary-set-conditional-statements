```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("\nset my_set:", my_set)
print("Explantion: A set stores only unique elements, so duplicate 'a' and 'i' values are removed. " 
      "the order is not guaranteed.")
```

```python
try:
    my_set[4] = 's'
except TypeError as e:
    print("b. Error while changing my_set[4]:", e)
    print("Explanation: Sets are unordered and do not support indexing or item assignment.")
```
